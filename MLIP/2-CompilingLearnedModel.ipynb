```python
# Cell: Final Model Compilation in "learned_MACE"

import os
import yaml
import subprocess

# Path to the original base config
BASE_CFG_PATH = 'config/config-02.yml'
with open(BASE_CFG_PATH) as f:
    base_cfg = yaml.safe_load(f)

# 1️⃣ Create the final model folder
learned_dir = "learned_MACE"
os.makedirs(learned_dir, exist_ok=True)

# 2️⃣ Define your final hyperparameters (override as needed)
#    Here’s an example—adjust num_interactions, num_channels, etc.
learned_cfg = dict(base_cfg,
    num_interactions = 3,
    num_channels     = 64,
    max_L            = 3,
    correlation      = 3,
    max_num_epochs   = 25,
    train_file       = 'data/solvent_xtb_train_4000.xyz',
    # point all outputs into learned_MACE
    model_dir        = learned_dir,
    log_dir          = learned_dir,
    checkpoints_dir  = learned_dir,
    results_dir      = learned_dir, 
    name = "mace_learned_N4000"
)

# 3️⃣ Write out the YAML config
learned_cfg_path = os.path.join(learned_dir, "learned_config.yml")
with open(learned_cfg_path, "w") as f:
    yaml.dump(learned_cfg, f)

print(f"✅ Wrote final config to {learned_cfg_path} \n\n\n\n")

# 4️⃣ Run training (will produce final model & compiled model in learned_MACE/)
!mace_run_train --config learned_MACE/learned_config.yml
```
