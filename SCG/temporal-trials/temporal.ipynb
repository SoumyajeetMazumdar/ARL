```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.fft import ifft, fftshift, fftfreq
from scipy.interpolate import interp1d
from sklearn.preprocessing import MinMaxScaler

# ---------------------------
# 1. Load spectrum from CSV
# ---------------------------
# Replace with your actual file path
file_path = 'mean_bk7_spectrum.csv' 
df = pd.read_csv(file_path)
df.columns = ['Wavelength(nm)', 'Intensity']

# Normalize intensity between 0 and 1
x_nm = df['Wavelength(nm)'].values
y = df['Intensity'].values
y = MinMaxScaler().fit_transform(y.reshape(-1, 1)).flatten()

plt.plot(x_nm,y)
plt.title("Original Spectrum, scaled")
```

```python
# ---------------------------
# 2. Apply Super-Gaussian Taper
# ---------------------------
from cProfile import label

from matplotlib.lines import lineStyles

def supergaussian(x, center, width, order, amplitude=1):
    return amplitude * np.exp(-((x - center) / (width / 2)) ** (2 * order))


center=675
width=800
order=10 # definition of the supergaussian

supergauss = supergaussian(x_nm, center, width, order)
tapered = y * supergauss

plt.plot(x_nm, supergauss, label="Supergaussian window")
plt.plot(x_nm, y, label="Original Spectrum")
plt.plot(x_nm, tapered, label = "Tapered spectrum", linestyle="--")
plt.grid(1)
plt.yscale("log")
plt.ylim(1e-4, 1)
plt.title(f"Remove noise using a SuperGaussian window\n Order:{order}, Width: {width}")
plt.legend()
```

```python
# ---------------------------
# 3. Convert to Angular Frequency
# ---------------------------
c_nm_fs = 299792458 * 1e-6  # speed of light in nm/fs
omega = 2 * np.pi * c_nm_fs / x_nm  # angular frequency in rad/fs

# Sort spectrum for FFT compatibility
sorted_idx = np.argsort(omega)
omega_sorted = omega[sorted_idx]
tapered_sorted = tapered[sorted_idx]

# Interpret tapered spectrum as E(ω) (field, not intensity)
# spectrum is intensity, convert to field amplitude
E_omega_pos = np.sqrt(tapered_sorted).astype(np.complex128)


plt.plot(omega_sorted, E_omega_pos)
plt.xlabel("$\omega$ - rad/fs")
plt.ylabel("Intensity")
plt.title(f"Conversion of x-axis to $\omega$-domain")
```

```python
# Apply jacobian for conversion from intensity to field amplitude
# E(ω) = sqrt(I(ω) * dω/dt)
domega_dt = np.gradient(omega_sorted)  # dω/dt
E_omega = E_omega_pos * np.sqrt(np.abs(domega_dt))  

plt.plot(omega_sorted, np.abs(E_omega))
plt.xlabel("$\omega$ - rad/fs")
plt.ylabel("Field Amplitude")
plt.title(f"Field Amplitude Spectrum after Super-Gaussian Tapering")
```

```python
# apply interpolation to equidistant frequency grid
# Interpolate to equidistant frequency grid
df_sorted = pd.DataFrame({'frequency_Hz': omega_sorted, 'intensity_corrected': np.abs(E_omega)})
df_sorted = df_sorted.sort_values(by='frequency_Hz')
original_freqs = df_sorted["frequency_Hz"].values
original_intensities = df_sorted["intensity_corrected"].values
uniform_freqs = np.linspace(original_freqs.min(), original_freqs.max(), len(original_freqs))    

# Use cubic interpolation
interpolator = interp1d(original_freqs, original_intensities, kind='cubic')
uniform_intensities = interpolator(uniform_freqs)

# scale intensities to 0-1 range
uniform_intensities = MinMaxScaler().fit_transform(uniform_intensities.reshape(-1, 1)).flatten()

plt.plot(uniform_freqs, uniform_intensities)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Intensity")
plt.title("Interpolated Spectrum on Equidistant Frequency Grid")
```

```python
# 4. Add zero padding to the right side of the positive frequencies (before adding negative frequencies)

zero_padding_factor = 10  # Increase this factor for more zero padding
num_points = len(uniform_freqs) * zero_padding_factor
uniform_freqs_padded = np.linspace(uniform_freqs.min(), uniform_freqs.max(), num_points)

uniform_intensities_padded = np.concatenate([
    np.zeros(num_points, dtype=complex),
    np.sqrt(tapered_sorted),  # using field amplitude, not intensity
    np.zeros(num_points, dtype=complex)
])



plt.plot()

```

```python
# ---------------------------
# 4. Zero-Padding + Hermitian Symmetric Spectrum + Plots
# ---------------------------

# Choose padding factor to control temporal resolution
padding_factor =  2
n_pad = len(E_omega_pos) * (padding_factor - 1) // 2

# Zero-pad the spectrum (before and after)
E_omega_padded = np.concatenate([
    np.zeros(n_pad, dtype=complex),
    np.sqrt(tapered_sorted),  # using field amplitude, not intensity
    np.zeros(n_pad, dtype=complex)
])

# Create Hermitian symmetric spectrum
E_omega_full = np.concatenate([
    E_omega_padded,
    np.conj(E_omega_padded[::-1])
])

# For plotting: angular frequency axis for padded and full spectrum
domega = np.mean(np.diff(omega_sorted))
omega_padded = np.linspace(
    omega_sorted.min() - domega * n_pad,
    omega_sorted.max() + domega * n_pad,
    len(E_omega_padded)
)
omega_full = np.concatenate([omega_padded, -omega_padded[::-1]])

# --- Plot the padded and Hermitian-symmetric spectrum ---
fig, axs = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

axs[0].plot(omega_padded, np.real(E_omega_padded), label='Re[E(ω)] (Padded)')
axs[0].plot(omega_padded, np.imag(E_omega_padded), label='Im[E(ω)] (Padded)', linestyle='--')
axs[0].set_title('Zero-Padded Field Spectrum (Before Hermitian Symmetry)')
axs[0].set_ylabel('Amplitude')
axs[0].grid(True)
axs[0].legend()

axs[1].plot(omega_full, np.real(E_omega_full), label='Re[E(ω)] (Hermitian)', alpha=0.9)
axs[1].plot(omega_full, np.imag(E_omega_full), label='Im[E(ω)] (Hermitian)', linestyle='--', alpha=0.9)
axs[1].set_title('Hermitian-Symmetric Spectrum for IFFT')
axs[1].set_xlabel('Angular Frequency ω (rad/fs)')
axs[1].set_ylabel('Amplitude')
axs[1].grid(True)
axs[1].legend()

plt.tight_layout()
plt.show()
```

```python
# ---------------------------
# 5. Implementing the actual IFFT
# # ---------------------------

# creating the time axis of the IFFT
n = len(E_omega_full) # number of time bins same as frequency bins
domega = np.mean(np.diff(omega_sorted)) # takes the mean del-omega to create del-t
t_fs = fftshift(fftfreq(n, domega / (2 * np.pi)))

# calculating the IFFT on the padded E(omega)
E_t = fftshift(ifft(E_omega_full))
I_t = np.abs(E_t)
I_t /= np.max(I_t)  # normalize intensity

plt.plot(t_fs, np.real(E_t), label= "real E(t)")
plt.plot(t_fs, np.imag(E_t), label="imaginary E(t)")
plt.grid(1)
plt.title(f"Temporal Electric Field after IFFT")
plt.legend()
plt.xlabel("time-fs")
plt.ylabel("E(t) - a.u.")
plt.xlim(-5, 5)
plt.show()


plt.plot(t_fs, I_t)
plt.grid(1)
plt.title(f"Temporal Intensity, Normalized")
plt.legend()
plt.xlabel("time-fs")
plt.ylabel("Intensity-a.u.")
plt.xlim(-5, 5)
plt.show()
```

```python
# ---------------------------
# 6. Calculate FWHM
# ---------------------------

def compute_fwhm(t, intensity):
    intensity = intensity / np.max(intensity)
    half_max = 0.5
    indices = np.where(intensity >= half_max)[0]

    if len(indices) < 2:
        return None  # FWHM not well-defined

    t1 = t[indices[0]]
    t2 = t[indices[-1]]
    return t1, t2


# ---------------------------
# 7. Plot Temporal Profile, with FWHM
# ---------------------------

t1, t2 = compute_fwhm(t_fs, I_t)
fwhm = t2-t1

plt.figure(figsize=(12, 6))
plt.plot(t_fs, I_t, label='Temporal Intensity |E(t)|²')
plt.axhline(0.5, color='gray', linestyle='--', label='Half Maximum')
plt.axvline(t1, color='red', linestyle='--', label=f'FWHM Start: {t1:.2f} fs')
plt.axvline(t2, color='green', linestyle='--', label=f'FWHM End: {t2:.2f} fs')
plt.title(f'Time-Domain Pulse with FWHM ≈ {fwhm:.2f} fs')
plt.xlabel('Time (fs)')
plt.ylabel('Normalized Intensity')
plt.xlim(-5, 5)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fftpack import fft, fftshift, fftfreq

# ------------------------------
# Step 1: Define time-domain field E(t)
# ------------------------------

# Physical and simulation parameters
c = 3e8  # speed of light (m/s)
lambda0 = 800e-9  # central wavelength in meters
f0 = c / lambda0  # central frequency in Hz
omega0 = 2 * np.pi * f0  # central angular frequency in rad/s

# # Time axis
# num_points = 2**14
# time_window = 50e-15  # total simulation window (s)
# t = np.linspace(-time_window/2, time_window/2, num_points)
# dt = t[1] - t[0]

# # Construct a non-Gaussian electric field as an example (arbitrary shaped pulse)
# # e.g., Gaussian with a weak second peak
# E_t = np.exp(-t**2 / (2 * (5e-15)**2)) + 0.5 * np.exp(-(t - 10e-15)**2 / (2 * (2e-15)**2))
# E_t *= np.cos(omega0 * t)  # carrier wave
t = t_fs
# E_t directly from above



# ------------------------------
# Step 2: Compute intensity I(t)
# ------------------------------

I_t = np.abs(E_t)**2  # temporal intensity profile

# ------------------------------
# Step 3: Define material and pulse parameters for SPM
# ------------------------------

n2 = 2.4e-20  # nonlinear refractive index for BK7 (m^2/W)
L = 6e-3      # propagation length in meters
pulse_energy = 8e-4  # pulse energy in joules (800 µJ)
beam_waist = 10e-6   # beam waist in meters

# Estimate peak intensity assuming temporal-spatial Gaussian profile
tau_fwhm = 1.75e-15
I0_est = 2 * pulse_energy / (np.pi * beam_waist**2 * tau_fwhm)

# Normalize intensity profile to match estimated I0 peak
I_t_scaled = I_t / np.max(I_t) * I0_est

# ------------------------------
# Step 4: Apply nonlinear phase shift
# ------------------------------

# Compute nonlinear phase shift at each time point
phi_nl = (2 * np.pi / lambda0) * n2 * L * I_t_scaled

# Apply SPM-induced phase modulation
E_out = np.abs(E_t) * np.cos(omega0 * t + phi_nl)  # purely real output

# ------------------------------
# Step 5: Compute and compare spectra
# ------------------------------
num_points = len(t)
dt = t[1] - t[0]
# Fourier transform
E_fft_in = fftshift(fft(E_t))
E_fft_out = fftshift(fft(E_out))
freq = fftshift(fftfreq(num_points, dt))  # frequency axis in Hz

plt.plot(freq, E_fft_in, label="input spectrum")
plt.plot(freq, E_fft_out, label="output spectrum")


# Spectral intensity
spectrum_in = np.abs(E_fft_in)**2
spectrum_out = np.abs(E_fft_out)**2
spectrum_in /= np.max(spectrum_in)
spectrum_out /= np.max(spectrum_out)

# Convert to wavelength for plotting
freq_pos = freq[freq > 0]
spectrum_in_pos = spectrum_in[freq > 0]
spectrum_out_pos = spectrum_out[freq > 0]
wavelength = c / (freq_pos)  # corrected mapping to center at 800 nm

# ------------------------------
# Step 6: Plot the results
# ------------------------------

plt.figure(figsize=(10, 5))
plt.plot(wavelength * 1e9, spectrum_in_pos, '--', label='Input Spectrum')
plt.plot(wavelength * 1e9, spectrum_out_pos, label='SPM Spectrum', linewidth=2)
plt.xlabel("Wavelength (nm)")
plt.ylabel("Normalized Spectral Intensity")
plt.title("SPM-Induced Spectral Broadening from Arbitrary E(t)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
