```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image

def add_Scale_to_image(img, scale_bar_length_px, scale_bar_label, bar_height=20, bar_color='red'):
    """
    Adds a scale bar to the image.
    
    Parameters:
    - img: PIL Image object
    - scale_bar_length_px: Length of the scale bar in pixels
    - scale_bar_label: Label for the scale bar
    - bar_height: Height of the scale bar in pixels
    - bar_color: Color of the scale bar
    """
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.imshow(img)
    ax.axis('off')

    # Positioning the scale bar
    bar_x = img.width - scale_bar_length_px - 50
    bar_y = img.height - 200

    # Add scale bar rectangle
    scale_bar = patches.Rectangle((bar_x, bar_y), scale_bar_length_px, bar_height, color=bar_color)
    ax.add_patch(scale_bar)

    # Add scale bar label
    ax.text(
        bar_x + scale_bar_length_px / 2,
        bar_y + bar_height + 30,
        scale_bar_label,
        color=bar_color,
        fontsize=15,
        ha='center',
        va='top'
    )

    return fig, ax


fig, ax = add_Scale_to_image(
    cropped_image,
    scale_bar_length_px=134*5,
    scale_bar_label="25 µm",
    bar_height=20,
    bar_color='red'
)   

plt.savefig("optical_flake_with_scalebar.png", dpi=300, bbox_inches='tight', pad_inches=0)
plt.show()




# # === Load your cropped image ===

# img = cropped_image

# # === Plotting Setup ===
# fig, ax = plt.subplots(figsize=(8, 8))
# ax.imshow(img)
# ax.axis('off')

# # === Scale bar settings ===
# scale_bar_length_px = 134*5  # 134px per 10 µm
# scale_bar_label = "50 µm"
# bar_height = 20
# bar_color = 'red'

# # === Add scale bar rectangle ===
# bar_x = img.width - scale_bar_length_px - 50
# bar_y = img.height - 200
# scale_bar = patches.Rectangle((bar_x, bar_y), scale_bar_length_px, bar_height, color=bar_color)
# ax.add_patch(scale_bar)

# # === Add scale bar label ===
# # Adjusted label placement
# ax.text(
#     bar_x + scale_bar_length_px / 2,
#     bar_y + bar_height + 30,
#     "50 µm",
#     color='red',
#     fontsize=15,
#     ha='center',
#     va='top'
# )


# # === Save final figure ===
# plt.savefig("final_with_scalebar.png", dpi=300, bbox_inches='tight', pad_inches=0)
# plt.show()
```
